## Import Functions to Test

```python
from raag_midi_gen.model.components.output_layer import OutputLayer
```

## Import Helper Functions

```python
from raag_midi_gen.datasets import midi_files_dataset
from raag_midi_gen.utils.midi_utils import play_muspy_music
from raag_midi_gen.tokenizer import encode_input, decode_output

from raag_midi_gen.model.components.embedding_layer import NoteEventEmbedding
```

```python
import torch
import numpy as np
```

## Get Data Sample

```python
# test_clip_filename = 'Taraana - Antra 2.0.mid'        # EASY
test_clip_filename = 'Aeri Aali - Sthaayi 1.1_2.mid'  # HARD
```

```python
dataset = midi_files_dataset.get_dataset()
sample_midi = dataset[test_clip_filename][-1]
```

## Compute the Tensors to Feed Into the Output Layer

1. Compute an `input_embedding` from the encoding of the `sample_midi`

```python
embedding_layer = NoteEventEmbedding()
```

```python
input_encoding = encode_input(sample_midi)
input_embedding = embedding_layer(input_encoding)
```

2. Compute a `random_input` such that it matches the `sample_midi`s shape

```python
random_input = torch.randn_like(input_embedding)
```

## Pass Embedding Through Output Layer

We don't necessarily have to pass it through the transformer encoder layer first, because a transfomer encoder layer outputs tensors with exactly the same shape as its inputs.

```python
output_layer = OutputLayer()
```

1. `sample_midi`

```python
output__from__input_embedding = output_layer(input_embedding)
```

2. `random_input`

```python
output_from__random_input = output_layer(random_input)
```

## Decode the Output Layer

1. `sample_midi`

```python
play_muspy_music(decode_output(*output__from__input_embedding))
```

2. `random_input`

```python
play_muspy_music(decode_output(*output_from__random_input))
```
